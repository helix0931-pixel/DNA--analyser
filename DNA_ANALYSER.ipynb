Final version of DNA analyser

```python
def analyse(dna):
  dna = dna.upper().strip()
  valid = {"A","G","T","C"}
  if not (set(dna) <= valid):
    print("ERROR NOT VALID")
    return
  a_count = 0
  t_count = 0
  g_count = 0
  c_count = 0
  length = len(dna)
  print("the length of dna is,",length)
  for base in dna :
    if base == "A" :
      a_count = a_count + 1
    elif base == "T" :
      t_count = t_count + 1
    elif base =="G" :
      g_count = g_count + 1
    elif base == "C" :
      c_count = c_count + 1
  print("Adenine = ",a_count)
  print("thymine = ",t_count)
  print("guanine = ",g_count)
  print("cytosine = ",c_count)
#gc content
  gc_content = round((g_count + c_count) /length*100,2)
  print("GC content = ",gc_content,"%")
#reverse strand
  DNA_pairs = {"A":"T","T":"A","G":"C","C":"G"}
  rev_comp = ""
  for basepair in dna[::-1]:
    rev_comp += DNA_pairs[basepair]
  print("reverse complement =",rev_comp)
# rna transcription
  rna = dna.replace("T","U")
  print("rna = ",rna)
```

```python
analyse("rehagjheheao;")
```

```python
def read_fasta(filename):
  lines = []
  with open(filename) as f:
    for line in f:
      line = line.strip()
      if line.startswith(">"):
        continue
      lines.append(line)
    sequence = "".join(lines)
    return(sequence)
#quick test
print(read_fasta("sample.fasta"))
```

```python
analyse(read_fasta("sample.fasta"))
```

FINAL ULTIMATE VERSION

```python
choice = input("dna format (file/seq) = ")
if choice == "file":
  file_name = input("file name")
  analyse(read_fasta(file_name))
elif choice == "seq" :
  dna = input("enter sequence")
  analyse(dna)
else :
  print("NOT VALID RESPONSE")
```
